## Step 1: Reading the data in Python

```python
import numpy as np
import pandas as pd

df = pd.read_csv('Quora-Duplicate-question.csv')

print(df.shape)
print(df.columns)
df.head()
```

## Step 2: Defining the problem statement

```python
problem_statement = "Classify whether two Quora questions are duplicates or not"
print(problem_statement)
```

## Step 3: Identifying the target variable

```python
target_column = 'is_duplicate'
print("Target variable:", target_column)
df[target_column].value_counts()
```

## Step 4: Looking at the distribution of the target variable

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='is_duplicate', data=df)
plt.title('Distribution of Target Variable (is_duplicate)')
plt.show()
```

## Step 5: Basic data exploration

```python
print(df.describe())
print(df.isnull().sum())
```

## Step 6: Feature Engineering

```python
df['q1_len'] = df['question1'].apply(lambda x: len(str(x)))
df['q2_len'] = df['question2'].apply(lambda x: len(str(x)))
df['q1_num_words'] = df['question1'].apply(lambda x: len(str(x).split()))
df['q2_num_words'] = df['question2'].apply(lambda x: len(str(x).split()))
```

## Step 7: Visual Exploratory Data Analysis

```python
plt.figure(figsize=(12, 6))
sns.histplot(df['q1_len'], bins=50, color='blue', label='Question 1 Length', kde=True)
sns.histplot(df['q2_len'], bins=50, color='red', label='Question 2 Length', kde=True)
plt.legend()
plt.title('Distribution of Question Lengths')
plt.show()

plt.figure(figsize=(12, 6))
sns.histplot(df['q1_num_words'], bins=50, color='blue', label='Question 1 Word Count', kde=True)
sns.histplot(df['q2_num_words'], bins=50, color='red', label='Question 2 Word Count', kde=True)
plt.legend()
plt.title('Distribution of Question Word Counts')
plt.show()
```

## Step 8: Rejecting useless columns

```python
df_cleaned = df.drop(columns=['id', 'qid1', 'qid2'])
```

## Step 9: Feature selection based on data distribution

```python
numeric_df = df_cleaned.select_dtypes(include=[np.number])
correlation_matrix = numeric_df.corr()
print(correlation_matrix['is_duplicate'].sort_values(ascending=False))
```

## Step 10: Outlier treatment

```python
for col in ['q1_len', 'q2_len', 'q1_num_words', 'q2_num_words']:
    upper_limit = df_cleaned[col].quantile(0.99)
    lower_limit = df_cleaned[col].quantile(0.01)
    df_cleaned[col] = np.where(df_cleaned[col] > upper_limit, upper_limit, df_cleaned[col])
    df_cleaned[col] = np.where(df_cleaned[col] < lower_limit, lower_limit, df_cleaned[col])
```

## Step 11: Missing values treatment

```python
df_cleaned = df_cleaned.fillna('')
```

## Step 12: Visual correlation analysis

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df_cleaned.select_dtypes(include=[np.number]).corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

## Step 13: Statistical correlation analysis (feature selection)

```python
from sklearn.feature_selection import SelectKBest, f_classif

X = df_cleaned[['q1_len', 'q2_len', 'q1_num_words', 'q2_num_words']]
y = df_cleaned['is_duplicate']

selector = SelectKBest(score_func=f_classif, k='all')
selector.fit(X, y)

feature_scores = pd.DataFrame({'Feature': X.columns, 'Score': selector.scores_})
print(feature_scores.sort_values(by='Score', ascending=False))
```

## Step 14: Converting data to numeric for ML

```python
from sklearn.feature_extraction.text import CountVectorizer

sample_df = df_cleaned.sample(50000, random_state=42)

questions = list(sample_df['question1']) + list(sample_df['question2'])
cv = CountVectorizer(max_features=3000)
q1_arr, q2_arr = np.vsplit(cv.fit_transform(questions).toarray(), 2)

import numpy as np
X_numeric = np.hstack((q1_arr, q2_arr, sample_df[['q1_len', 'q2_len', 'q1_num_words', 'q2_num_words']].values))
y_numeric = sample_df['is_duplicate'].values
```

## Step 15: Sampling 

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

X_train, X_test, y_train, y_test = train_test_split(X_numeric, y_numeric, test_size=0.2, random_state=42)

model = RandomForestClassifier(n_estimators=50, n_jobs=-1)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
acc = accuracy_score(y_test, y_pred)
print(f"RandomForest Test Accuracy: {acc:.4f}")
```

## Step 16: Trying multiple regression (classification) algorithms

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(solver='liblinear', max_iter=500)
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)
print("LogisticRegression Test Accuracy:", accuracy_score(y_test, y_pred_lr))
```

```python
from sklearn.model_selection import cross_val_score, StratifiedKFold
from xgboost import XGBClassifier

kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

xgb = XGBClassifier(
    n_estimators=50,
    tree_method='gpu_hist',
    predictor='gpu_predictor',
    use_label_encoder=False,
    eval_metric='logloss'
)

scores = cross_val_score(xgb, X_numeric, y_numeric, cv=kfold, scoring='accuracy')
print("XGBoost 5-fold CV scores:", scores)
print("Mean CV accuracy:", np.mean(scores))

```
